```python
# Concept operations in Python
num1 = float(input("Enter the first number"))
num2 = float(input("Enter the second number"))
choice = input("Enter the operation you want to perform (+, -, *, /): ")
if choice == '+':
    print(num1 + num2)
elif choice == '-':
    print(num1 - num2)
elif choice == '*':
    print(num1 * num2)
elif choice == '/':
    print(num1 / num2)
else:
    print("Invalid choice")
```

```python
#  Concept of fibonacci series
num = int(input("Enter a number for factorial calculation: "))
factorial = 1
for i in range(1, num + 1):
    factorial *= i
print("The factorial of", num, "is", factorial)
```

```python
#Concept of fibonacci series
num = int(input("Enter the number of terms for the Fibonacci series: "))
a = 0
b = 1
print("Fibonacci series:")
for i in range(num):
    print(a, end=' ')
    a, b = b, a + b
```

```python
# Concept of dictionary in Python
train_details = {"Train_Name" : ['A' , 'B' , 'C' ],
           "Seats_Available" : [10 , 8 , 9 ],
           "Price" : [800 , 900 , 1000 ]}
for key, value in train_details.items():
    print(f'{key} and {value}')
train_details.update({"Km" : [432 , 234 , 234 ]})
for key, value in train_details.items():
    print(f'{key} and {value}')
```
