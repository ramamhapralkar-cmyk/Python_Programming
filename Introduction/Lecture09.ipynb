```python
class employee:
    #Define constructor
    def __init__(self,name,age,
                 department,salary):
      self.name = name
      self.age = age
      self.department = department
      self.salary = salary
      
      
    def employee_info(self):
        print("name: ",self.name)
        print("age: ",self.age)
        print("department: ",self.department)
        print("salary: ",self.salary)
        
        
        
class manager(employee):
           
    #Define constructor
    def __init__(self,name,age,
                    department,salary,designation):
            
        super().__init__(name,age,department,salary)
        self.designation = designation
            
    def info(self):    
         print(f'The Name of manager is {self.name}')
         print(f'The Designation is {self.designation}')
            
#Define object
emp1 = employee('Darshan',21,'HR',200000)
emp2 = manager('Rama',21,'IT',300000,'Admin')
```

```python
emp1.employee_info()
```

```python
emp2.info()
```
