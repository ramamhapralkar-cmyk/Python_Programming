```python
class motor: 
    
    def __init__(self,wheel,color,engine):
        self.wheel = wheel
        self.color = color
        self.engine = engine
        
    def sound(self):
        print('beep')
            
    
tata = motor(4,'black','petrol')
audi = motor(4,'blue','petrol')
bmw = motor(4,'red','petrol')
suzuki = motor(4,'black','diesel')
```

```python
class railone:
    
    def __init__(self,train_name,passanger_name,age,sex,seat_no):
      
        self.train_name = train_name
        self.passanger_name = passanger_name
        self.age = age
        self.sex = sex
        self.seat_no = seat_no
    
    def train_info(self):
        print(f'The Train Name is {self.train_name}')
    
    def passanger_info(self):
        print(f'The Passanger Name is {self.passanger_name}')
        print(f'The Age of passanger is {self.age}')      
        print(f'The sex of the passanger {self.sex}')
        print(f'The seat no. of Passanger is {self.seat_no}')
    
train_name = input('Enter the train name : ')
passanger_name = input('Enter the passanger name : ')
age = input('Enter the age : ')
sex = input('Enter the sex : ')
seat_no = input('Enter the seat_no : ')


Maininfo = railone(train_name,passanger_name,age,sex,seat_no)
```

```python
Maininfo.train_info()
```
