```python
# blueprint/ templates : class bank
class bank:
    
    #constructor/special function/special method
    def __init__(self,name):
        self.name = name      #attribute#1
        
    def emp(self):              #method
        name = self.name         #attribute#1  
        
    # Fefine object: sbi
    sbi = bank('india')
```
