```python
class bank: #parent class
    #define constructor 
    def _init_(self,acct_no,acct_holder,
                 age,gender,email_id,
                 address,mobile_no,user_id,passcode):

        self.acct_no=acct_no
        self.acct_holder=acct_holder
        self.age=age
        self.gender=gender
        self.email_id=email_id
        self.address=address
        self.mobile_no=mobile_no
        self.user_id=user_id
        self.__passcode=passcode

        
#define child class : saving_Acct 
class saving_acct(bank):
    #define constructor
    def _init_(self, interest_rate, available_balance):
        super()._init_(acct_no, acct_holder,
                        age, gender, 
                        email_id, address, 
                        mobile_no, user_id, passcode)
        
        self.interest_rate=interest_rate
        self.available_balance=available_balance 

    def credit(self):
        self.credit = int(input("Enter the amount to be credited: "))

        print(f"--------------------------------------------------------------")
        print(f"acct number: {self.acct_no}")
        print(f"acct holder: {self.acct_holder}")
        print(f"age: {self.age}")
        print(f"gender: {self.gender}")
        print(f"email id: {self.email_id}")
        print(f"address: {self.address}")
        print(f"mobile no: {self.mobile_no}")
        print(f"user id: {self.user_id}")
        print(f"passcode: {self.__passcode}")
        print(f"interest rate: {self.interest_rate}")
        print(f"available balance: {self.available_balance}")
        print(f"the amount credited: {self.credit}")
        print(f"new updated balance: {self.available_balance + self.credit}")
        print(f"credit amount: {self.credit}")
        print(f"--------------------------------------------------------------")
```

```python
acct_no = int(input("Enter account number: "))
acct_holder = input("Enter account holder name: ")
age = int(input("Enter age: "))
gender = input("Enter gender: ")
email_id = input("Enter email ID: ")
address = input("Enter address: ")
mobile_no = int(input("Enter mobile number: "))
user_id = input("Enter user ID: ")
passcode = input("Enter passcode: ")
interest_rate = int(input("Enter interest rate: "))
available_balance = int(input("Enter available balance: "))
```

```python
hdfc_1 = saving_acct(interest_rate, available_balance) # Object of child class
```
