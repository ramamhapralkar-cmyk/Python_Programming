```python
class railone:
  irctc = 'Indian Railway'   # Class Variable/Class Attribute

  # Define Constructor/ Special Method/ Special Function/ Instance Method
  def __init__(self,train_name,passanger_name,gender,age):
    self.train_name = train_name  # Instance Variable/Attribute1
    self.passanger_name = passanger_name # Instance Variable/Attribute2
    self.gender = gender  # Instance Variable/Attribute3
    self.age = age # Instance Variable/Attribute4

  # Define Class Method
  def passanger_info(self):
    self.address = input('Enter the address : ')  # local attribute
    print(f'Train Name: {self.train_name}')
    print(f'Passanger Name: {self.passanger_name}')
    print(f'Gender: {self.gender}')
    print(f'Age: {self.age}')
    print(f'Address: {self.address}')

  # Define Class Method
  def booking_info(self):
    self.booking_available = 10  # local attribute
    self.booking = int(input('Enter the booking : '))
    self.fare = 100
    self.choice = input('Enter the booking choice :').lower()

    if self.choice == 'firstclass':

      if (self.booking <= self.booking_available):
        print('Booking is Available')
        self.fare = self.fare * self.booking
        self.booking_available = self.booking_available - self.booking
        print(f'Train Name: {self.train_name}')
        print(f'Passanger Name: {self.passanger_name}')
        print(f'The fare is {self.fare}')
        print(f'Booking Available: {self.booking_available}')
      else:
        print('Booking is Not Available')


train_name = input('Enter the Train Name : ')
passanger_name = input('Enter the Passanger Name : ')
gender = input('Enter the Gender : ')
age = int(input('Enter the Age : '))

railone_1 = railone(train_name,passanger_name,gender,age)
```

```python
railone_1.booking_info()
```

```python
#homework-Bank

class icici:
    bank_name = 'ICICI Bank'   #class variable
    
# define constructor
    def __init__(self, account_holder,
                 age, gender, account_no):
        
        self.account_holder = account_holder      # Instance Variable1
        self.age = age                            # Instance Variable2
        self.gender = gender                      # Instance Variable3
        self.account_no = account_no              # Instance Variable4

    # Define class method
    def customer_info(self):
        self.address = input('Enter the address: ')   # local attribute

        print(f'Bank Name: {self.bank_name}')
        print(f'Account Holder: {self.account_holder}')
        print(f'Age: {self.age}')
        print(f'Gender: {self.gender}')
        print(f'Account Number: {self.account_no}')
        print(f'Address: {self.address}')

    # Define class method
    def deposit_info(self):
        self.balance = 5000  #local attribute

        self.deposit = int(input('Enter the deposit amount: '))

        self.balance = self.balance + self.deposit

        print(f'Account Holder: {self.account_holder}')
        print(f'Deposit Amount: {self.deposit}')
        print(f'Total Balance: {self.balance}')

    # Define class method
    def withdraw_info(self):
        self.withdraw = int(input('Enter the withdrawal amount: '))

        if self.withdraw <= self.balance:
            self.balance = self.balance - self.withdraw

            print('Withdrawal Successful')
            print(f'Account Holder: {self.account_holder}')
            print(f'Withdrawal Amount: {self.withdraw}')
            print(f'Remaining Balance: {self.balance}')

        else:
            print('insufficient Balance')


account_holder = input('Enter the Account Holder Name: ')
age = int(input('Enter the Age: '))
gender = input('Enter the Gender: ')
account_no = int(input('Enter the Account Number: '))


icici_1 = icici(account_holder, age, gender, account_no)


```

```python
icici_1.customer_info()
```
