```python
marks = float(input("Enter the marks"))
if marks >= 90:
    print(" Grade A+ ")
elif marks >= 75:
    print( " Grade A ")
elif marks >= 60:
    print( " Grade B ")
elif marks >= 40:
    print( " Grade c")
else:
    print("Fail")
    
    
    
```
